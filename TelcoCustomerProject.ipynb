------- Solving KPIs ----------

```python
engine1 = create_engine(f"mysql+pymysql://root:nopassword@localhost:3306/telco")
```

```python
# setting a function to remove engine typing
def run_query(sql):
    return pd.read_sql(sql, engine1)
```

```python
# 1. Churn by Contract Type: Compares the churn volume of month-to-month contracts against 1-year and 2-year contracts.

churnVolume = run_query("""SELECT 
    Contract, 
    COUNT(Churn) AS LeaveCount, 
    ROUND((COUNT(Churn) * 100.0) / SUM(COUNT(Churn)) OVER(), 2) AS PercentChurn
FROM telco.telco_customers
WHERE Churn = 1
GROUP BY Contract""")

# SELECT Contract, COUNT(Churn) AS LeaveCount, (COUNT(Churn)*100)/SUM(COUNT(Churn)) OVER() AS PercentChurn
# from telco.telco_customers
#where Churn = 1
# group by Contract;

# Adding the OVER() clause tells the database to calculate the SUM across the entire dataset returning from the WHERE filter, bypassing the local GROUP BY limitation.
```

```python
# Exporting to csv -  the above result table of calculation.
churnVolume.to_csv('churnVolume.csv', index=False)
```

```python
churnVolume
```

```python
import matplotlib.pyplot as plt
```

```python
churnVolume = churnVolume.sort_values('LeaveCount', ascending=False)

plt.figure(figsize=(8, 5))

# Create the bar chart
bars = plt.bar(churnVolume['Contract'], churnVolume['LeaveCount'], color='#1f77b4', edgecolor='black')

# Add the percentages as data labels on top of each bar
for bar, percentage in zip(bars, churnVolume['PercentChurn']):
    yval = bar.get_height()
    # Position the text slightly above the bar
    plt.text(bar.get_x() + bar.get_width()/2, yval + 20, 
             f'{percentage}%', ha='center', va='bottom', fontsize=11, fontweight='bold')

# Formatting the chart to look professional
plt.title('Churn Volume by Contract Type', fontsize=14, pad=15)
plt.xlabel('Contract Type', fontsize=12)
plt.ylabel('Number of Churned Customers', fontsize=12)

# Remove top and right borders for a cleaner "dashboard" look
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

# Display the chart
plt.tight_layout()
plt.show()
```

```python
# 2. Churn by Tenure: Groups customers by the number of months they stayed (e.g., 0-6 months, 6-12 months) to identify the highest risk period.

churnByTenure = run_query("""SELECT
	CASE
		WHEN Tenure <= 6 THEN '0-6 Months'
		WHEN Tenure > 6 AND Tenure <= 12 THEN '6-12 Months'
		WHEN Tenure >12 AND Tenure <= 18 THEN '12-18 Months'
		WHEN Tenure >18 AND Tenure <= 24 THEN '18-24 Months'
		WHEN Tenure >24 AND Tenure <= 30 THEN '24-30 Months'
		WHEN Tenure >30 AND Tenure <= 36 THEN '30-36 Months'
		WHEN Tenure >36 AND Tenure <= 42 THEN '36-42 Months'
		WHEN Tenure >42 AND Tenure <= 48 THEN '42-48 Months'
		WHEN Tenure >48 AND Tenure <= 54 THEN '48-54 Months'
		WHEN Tenure >54 AND Tenure <= 60 THEN '54-60 Months'
		WHEN Tenure >60 AND Tenure <= 66 THEN '60-66 Months'
		WHEN Tenure >66 AND Tenure <= 72 THEN '66-72 Months'
	END AS 'Months_Interval',
	
	COUNT(Churn) As Churn
FROM telco.telco_customers
where Churn=1
group by Months_Interval
order by MIN(Tenure)""")
```

```python
churnByTenure
```

```python
# Exporting to csv -  the above result table of calculation.
churnByTenure.to_csv('churnByTenure.csv', index=False)
```

```python
plt.figure(figsize=(10, 6))

# Create the bar chart
bars = plt.bar(churnByTenure['Months_Interval'], churnByTenure['Churn'], color='#ff7f0e', edgecolor='black')

# Add the volume count as data labels on top of each bar
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 15, 
             int(yval), ha='center', va='bottom', fontsize=10)

# Formatting the chart
plt.title('Churn Volume by Tenure Interval', fontsize=14, pad=15)
plt.xlabel('Tenure Groups', fontsize=12)
plt.ylabel('Number of Churned Customers', fontsize=12)

# Rotate X-axis labels by 45 degrees so they fit cleanly
plt.xticks(rotation=45, ha='right')

# Remove top and right borders
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

# Display the chart
plt.tight_layout()
plt.show()
```

```python
# 3. Monthly Charges vs. Retention: Compares the average monthly bill of customers who left versus those who stayed.

avg_billingByChurn = run_query("""SELECT 
	CASE
		WHEN churn = 1 THEN 'Churn Customers'
		WHEN churn = 0 THEN 'Non Churn Customers'
	END AS 'Customer_Type',
	
	ROUND(AVG(MonthlyCharges), 2) as Avg_Monthly_Billing 
	
FROM telco.telco_customers
WHERE Churn = 1 or Churn = 0
GROUP BY Customer_Type""")

# The output reveals a major business insight for your portfolio: customers who leave are paying approximately $13 more per month on average than those who stay.
```

```python
avg_billingByChurn
```

```python
# Exporting to csv -  the above result table of calculation.
avg_billingByChurn.to_csv('avg_billingByChurn.csv', index=False)
```

```python
plt.figure(figsize=(7, 5))

# Using distinct colors to contrast retained (green) vs. churned (red)
colors = ['#2ca02c', '#d62728'] 

# Create the bar chart
bars = plt.bar(avg_billingByChurn['Customer_Type'], avg_billingByChurn['Avg_Monthly_Billing'], color=colors, edgecolor='black', width=0.5)

# Add the exact dollar amounts on top of the bars
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 1.5, 
             f'${yval}', ha='center', va='bottom', fontsize=11, fontweight='bold')

# Formatting the chart
plt.title('Average Monthly Billing: Retained vs. Churned', fontsize=14, pad=15)
plt.ylabel('Average Monthly Bill ($)', fontsize=12)

# Remove top and right borders for a clean look
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
# 4. Impact of Tech Support: Contrasts the churn rate of customers who use tech support services against those who do not.

churnByTechSupport = run_query("""SELECT 
	CASE
		WHEN TechSupport = "Yes" THEN 'TechSupport'
		WHEN TechSupport = "No" THEN 'Non TechSupport'
		ELSE 'No Internet'
	END AS 'UsingTechSupport',
	
	ROUND(AVG(Churn)* 100, 2) as Churn_Percentage
	
FROM telco.telco_customers

GROUP BY UsingTechSupport""")
```

```python
# Exporting to csv -  the above result table of calculation.
churnByTechSupport.to_csv('churnByTechSupport.csv', index=False)
```

```python
plt.figure(figsize=(8, 5))

# Create the bar chart
bars = plt.bar(churnByTechSupport['UsingTechSupport'], churnByTechSupport['Churn_Percentage'], color='#9467bd', edgecolor='black', width=0.6)

# Add the percentage data labels on top of each bar
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 1, 
             f'{yval}%', ha='center', va='bottom', fontsize=11, fontweight='bold')

# Formatting the chart
plt.title('Churn Rate by Tech Support Usage', fontsize=14, pad=15)
plt.xlabel('Tech Support Status', fontsize=12)
plt.ylabel('Churn Rate (%)', fontsize=12)

# Remove top and right borders for a clean look
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
# 5. Churn by Service Type: Breaks down churn percentages across different internet setups (Fiber Optic vs. DSL).

churnByServiceType = run_query("""SELECT InternetService, ROUND(AVG(Churn)*100, 2) AS ChurnPercentage
                                    FROM telco.telco_customers
                                    GROUP BY InternetService""")

# Fiber Optic customers are churning at an alarming rate of nearly 42%, which is more than double the churn rate of DSL customers
```
